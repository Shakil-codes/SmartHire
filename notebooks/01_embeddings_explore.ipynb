In this notebook, we explore how text embeddings capture contextual meaning and how cosine similarity is used to find matching jobs beyond simple keyword matching.

```python
import sys
sys.path.append('..')
import numpy as np
from src.search.embed import get_embedding_function
```

## 1. Initialize the Embedding Model
SmartHire GenAI uses Google Gemini's `text-embedding-004` (with an automatic local fallback when developing offline).

```python
embed_fn = get_embedding_function()
print(f"Active Embedding Function: {type(embed_fn).__name__}")
```

## 2. Test Semantic Vectors vs Pure Keywords
Let us embed three distinct sentences:
1. **Sentence A**: Data Science, machine learning, and statistical modeling in Python
2. **Sentence B**: AI predictive analytics, neural networks, and algorithms (Semantically very similar, different words)
3. **Sentence C**: Traditional wood-fired pizza dough recipe with cheese (Completely unrelated domain)

```python
texts = [
    "Data Science, machine learning, and statistical modeling in Python",
    "AI predictive analytics, neural networks, and algorithms",
    "Traditional wood-fired pizza dough recipe with cheese"
]
vectors = embed_fn.embed_documents(texts)
print(f"Successfully generated {len(vectors)} vectors of dimension {len(vectors[0])}")
```

## 3. Cosine Similarity Calculation
We calculate the cosine similarity between the vector embeddings:
$$\text{Cosine Similarity}(\vec{u}, \vec{v}) = \frac{\vec{u} \cdot \vec{v}}{\|\vec{u}\| \|\vec{v}\|}$$

```python
def cosine_similarity(v1, v2):
    vec1 = np.array(v1)
    vec2 = np.array(v2)
    return float(np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2)))

sim_AB = cosine_similarity(vectors[0], vectors[1])
sim_AC = cosine_similarity(vectors[0], vectors[2])

print(f"Similarity (Data Science vs AI Analytics): {sim_AB:.4f}")
print(f"Similarity (Data Science vs Pizza Recipe):  {sim_AC:.4f}")
assert sim_AB > sim_AC, "Semantic check failed: Similar concepts must score higher than unrelated ones!"
print("Semantic relationship successfully verified!")
```
